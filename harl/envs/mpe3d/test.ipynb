```python
env = MPE3DContinuousEnv(render_mode="rgb_array",
                             n_ugv=4, n_evaders=1, physics_hz=30,
                             bev_range_min=25.0, bev_range_max=120.0, bev_margin=6.0,
                             planner_hz=2,
                             record_video=True, video_file="./videos/mpe3d_demo1.mp4",
                             voxel_z_stride=1, voxel_obstacle_alpha=0.35)
obs, _ = env.reset(seed=7)
```

```python
for t in range(100):
    a = np.zeros(2*env.n_ugv, np.float32)
    # simple goal-chasing for preview (replace with MARL)
    for i,g in enumerate(env._goals):
        p = env.ugvs[i].pos[:2]
        tgt = np.array([g.pos_xy[0], g.pos_xy[1]], np.float32)
        dir_ = normalize_vec(tgt - p)
        a[2*i:2*i+2] = dir_ * env.ugvs[i].kin.max_speed * 0.7
    obs, r, term, trunc, info = env.step(a)
    # if t % 30 == 0:
    #     env.render()
    if term or trunc:
        break
env.close()
```
